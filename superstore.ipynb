```python
df.columns = df.columns.str.lower().str.replace(' ','_')
df.columns = df.columns.str.lower().str.replace('-','_')
df.columns
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
df.info()
```

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
df.head()
```

```python
df['shipping_days'] = (
    df['ship_date'] - df['order_date']
).dt.days
df[['order_date', 'ship_date', 'shipping_days']].head()
```

CATEGORY ANALYSIS

Which product category generates the most sales and profit?

```python
category_analysis = df.groupby('category').agg(
    total_sales=('sales','sum'),
    total_profit = ('profit','sum'),
    total_quantity = ('quantity','sum')
).sort_values('total_sales',ascending=False)

category_analysis
```

Profit Margin

```python
category_analysis['profit_margin'] = (
    round(category_analysis['total_profit']/
    category_analysis['total_sales']*100, 2)
)
category_analysis

#Furniture generates substantial revenue but converts very little of that revenue into profit.
```

```python
import matplotlib.pyplot as plt
category = ['Technology','Furniture','Office Supplies']
plt.bar(category,category_analysis['total_sales'])
plt.title('category to sales')
plt.show()
```

SUBCATEGORY ANALYSIS

```python
subcategory_analysis = df.groupby('sub_category').agg(
    total_sales = ('sales','sum'),
    total_profit = ('profit','sum'),
    total_quantity = ('quantity','sum')
)
subcategory_analysis
```

```python
subcategory_analysis['profit_margin'] = (
    subcategory_analysis['total_profit']/
    subcategory_analysis['total_sales']*100    
)

subcategory_analysis.sort_values('total_profit',ascending=False)

#Labels has the highest margin (44.42%), while Copiers generates the highest absolute profit ($55,618).
#It generates more than $206K in sales, but loses almost $17.7K.
```

DISCOUNT ANALYSIS ----------------------------

```python
discount_analysis = df.groupby('discount').agg(
    total_sales= ('sales','sum'),
    total_profit= ('profit','sum'),
    order_count = ('order_id','count'),
).reset_index()

discount_analysis
```

Discount vs Profit

```python
discount_analysis['profit_margin'] = (
    discount_analysis['total_profit']/
    discount_analysis['total_sales']*100
)

discount_analysis

#So the data strongly suggests that heavy discounting is associated with substantial losses.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.barplot(data = discount_analysis, x = 'discount', y = 'total_profit')
plt.title('Discount VS Profit')
plt.show()
```

```python
region_analysis['profit_margin'] = (
    region_analysis['total_profit']/
    region_analysis['total_sales']*100
)
region_analysis
```

State Analysis

```python
state_analysis = df.groupby('state').agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    total_quantity=('quantity', 'sum')
).reset_index().sort_values('total_sales', ascending=False)

state_analysis['profit_margin'] = (
    state_analysis['total_profit'] /
    state_analysis['total_sales'] * 100
)

state_analysis
```

Top & Bottom Products

```python
product_analysis = df.groupby('product_name').agg(
    total_sales=('sales','sum'),
    total_profit=('profit','sum'),
    total_quantity=('quantity','sum')
).reset_index()

product_analysis['profit_margin'] = (
    product_analysis['total_profit'] /
    product_analysis['total_sales'] * 100
)
product_analysis
```

Top 10 products by profit

```python
top_products = product_analysis.sort_values('total_profit', ascending=False).head(10)
top_products
```

Bottom 10 products by profit

```python
bottom_products = product_analysis.sort_values('total_profit', ascending=True).head(10)
bottom_products
```

discount vs profit at the product level.


```python
product_discount_analysis = df.groupby('product_name').agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    avg_discount=('discount', 'mean'),
    total_quantity=('quantity', 'sum')
).reset_index()

product_discount_analysis['profit_margin'] = (
    product_discount_analysis['total_profit'] /
    product_discount_analysis['total_sales'] * 100
)

product_discount_analysis.sort_values(
    'total_profit',
    ascending=True
).head(10)
```

Shipping Performance


```python
shipping_analysis = df.groupby('ship_mode').agg(
    avg_shipping_days=('shipping_days', 'mean'),
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    order_count=('order_id', 'count')
)

shipping_analysis

#Standard Class is the dominant shipping mode, accounting for the highest sales, profit, and order volume despite having the longest average shipping time of approximately 5 days. Same Day shipping is the fastest option but represents a much smaller portion of orders.
```
